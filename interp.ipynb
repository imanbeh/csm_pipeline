```python
import numpy as np
def func(x, y):
    return x*(1-x)*np.cos(4*np.pi*x) * np.sin(4*np.pi*y**2)**2
```

```python
grid_x, grid_y = np.mgrid[0:1:100j, 0:1:200j]
```

```python
grid_y
```

```python
rng = np.random.default_rng()
points = rng.random((1000, 2)) # 2d data
values = func(points[:,0], points[:,1]) # new dimension, density map, 1d array
```

```python
values.shape
```

```python
from scipy.interpolate import griddata
grid_z0 = griddata(points, values, (grid_x, grid_y), method='nearest')
grid_z1 = griddata(points, values, (grid_x, grid_y), method='linear')
grid_z2 = griddata(points, values, (grid_x, grid_y), method='cubic')
```

```python
import matplotlib.pyplot as plt
plt.subplot(221)
plt.imshow(func(grid_x, grid_y).T, extent=(0,1,0,1), origin='lower')
plt.plot(points[:,0], points[:,1], 'k.', ms=1)
plt.title('Original')
plt.subplot(222)
plt.imshow(grid_z0.T, extent=(0,1,0,1), origin='lower')
plt.title('Nearest')
plt.subplot(223)
plt.imshow(grid_z1.T, extent=(0,1,0,1), origin='lower')
plt.title('Linear')
plt.subplot(224)
plt.imshow(grid_z2.T, extent=(0,1,0,1), origin='lower')
plt.title('Cubic')
plt.gcf().set_size_inches(6, 6)
plt.show()
```

Rect Bivarotate Spine

```python
import numpy as np
from scipy.interpolate import RectBivariateSpline
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 4, 13)
y = np.array([0, 2, 3, 3.5, 3.75, 3.875, 3.9375, 4])
X, Y = np.meshgrid(x, y) #coordinate matrix
Z = np.sin(np.pi * X / 2) * np.exp(Y / 2) #values at each point
```

```python
type((x,y))
```

```python
x2 = np.linspace(0, 4, 65)
y2 = np.linspace(0, 4, 65)
f = RectBivariateSpline(x, y, Z.T) # passing a xy grid with values of Z through interpolation function
Z2 = f(x2, y2).T    # interpolating values at new x2 and y2 values
```

```python
fig, ax = plt.subplots(nrows=1, ncols=2)
ax[0].pcolormesh(X, Y, Z)
ax[0].set_ylim(0, 4)

X2, Y2 = np.meshgrid(x2, y2)
ax[1].pcolormesh(X2, Y2, Z2)
ax[1].set_ylim(0, 4)

plt.show()
```
